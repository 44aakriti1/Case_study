```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("chik0di/health-and-lifestyle-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# See what files are in the folder
print(os.listdir(path))

# Load the CSV (replace filename with whatever shows up above)
df = pd.read_csv(os.path.join(path, "health_lifestyle_dataset.csv"))
print(df.head())
```

```python
from sklearn.preprocessing import LabelEncoder

le  = LabelEncoder()
df['gender_encoded'] = le.fit_transform(df['gender'])
print("Target Encoding Mapping:")
for i, item in enumerate(le.classes_):
    print(f"{item} --> {i}")

df.drop('gender',axis=1,inplace = True)
```

```python
correlation_matrix = df.corr()
print(correlation_matrix)
```

```python
plt.figure(figsize=(15,15)) # Adjust the figure size if needed
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
```
